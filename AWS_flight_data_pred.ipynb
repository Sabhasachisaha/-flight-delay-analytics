```python
BUCKET_NAME = 'showmo-project-2026'
RAW_DATA_PATH = r'C:\Users\User\OneDrive - Clemson University\flight-delay-analytics\data\flights_raw.csv'
S3_KEY = 'flights_raw.csv'

# DON'T create a new s3_client here
# Just use the one from the credentials cell

try:
    s3_client.upload_file(RAW_DATA_PATH, BUCKET_NAME, S3_KEY)
    print(f"✓ Successfully uploaded {S3_KEY} to {BUCKET_NAME}")
except FileNotFoundError:
    print(f"✗ Error: File not found at {RAW_DATA_PATH}")
except Exception as e:
    print(f"✗ Error uploading file: {e}")
```

```python
PREDICTIONS_PATH = r'C:\Users\User\OneDrive - Clemson University\flight-delay-analytics\predictions.csv'
S3_KEY = 'predictions.csv'

try:
    s3_client.upload_file(PREDICTIONS_PATH, BUCKET_NAME, S3_KEY)
    print(f"✓ Successfully uploaded {S3_KEY} to {BUCKET_NAME}")
except FileNotFoundError:
    print(f"✗ Error: File not found at {PREDICTIONS_PATH}")
except Exception as e:
    print(f"✗ Error uploading file: {e}")
```

```python
import pandas as pd

S3_KEY = 'predictions.csv'

try:
    response = s3_client.get_object(Bucket=BUCKET_NAME, Key=S3_KEY)
    df = pd.read_csv(response['Body'])
    
    print("=" * 60)
    print("PREDICTIONS SUMMARY")
    print("=" * 60)
    print(f"\nTotal predictions: {len(df)}")
    print(f"Correct: {df['correct'].sum()} ({df['correct'].mean()*100:.2f}%)")
    print(f"Incorrect: {(1 - df['correct']).sum()} ({(1 - df['correct']).mean()*100:.2f}%)")
    print(f"\n✓ Successfully read {S3_KEY} from {BUCKET_NAME}")
    print("=" * 60)
    
except Exception as e:
    print(f"✗ Error: {e}")
```
